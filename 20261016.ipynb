```python
#정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요.
n=int(input("정수를 입력하세요."))
for i in range(1, 10):
  print(f"{n}*{i}={n*i}")
```

```python
#2단부터 9단까지 구구단을 7 x 1 = 7 형식으로 출력하세요.
for i in range(2, 10):
  for j in range(1, 10):
    print(i,"x",j,"=",i*j)
```

```python
#정수n을 입력받아 n의 약수의 개수를 출력하고, 약수가 2개이면 "소수입니다."를 함께 출력하세요.
n=int(input())
count=0

for i in range(1, n+1):
  if n%i==0:
    count+=1

print(f"약수의 개수:{count}")
if count==2:
  print("소수입니다.")
```

```python
#while문으로 1부터 100까지의 합을 출력하시오.
sum=0
i=1
while i<=100:
  sum+=i
  i+=1
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요.
sum=0
while True:
  n=int(input("양의 정수를 입력하세요."))
  if n==0:
    break
  sum+=n
print(sum)
```

```python
#1부터 20까지의 수 중 짝수만 담은 리스트를 만들어 출력하세요.
evens=[]
for i in range(1,21):
  if i%2==0:
   evens.append(i)
print(evens)
evens.pop(7)
print(evens)
```
