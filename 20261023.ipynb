```python
# 국가명과 수도를 저장하는 딕셔너리
capitals = {
    "대한민국": "서울",
    "일본": "도쿄",
    "중국": "베이징",
    "미국": "워싱턴 D.C.",
    "영국": "런던",
    "프랑스": "파리",
    "독일": "베를린",
    "이탈리아": "로마",
    "캐나다": "오타와",
    "호주": "캔버라",
    "베트남": "하노이",
    "태국": "방콕",
    "인도": "뉴델리",
    "스페인": "마드리드",
    "브라질": "브라질리아",
    "멕시코": "멕시코시티",
    "이집트": "카이로",
    "터키": "앙카라",
    "러시아": "모스크바",
    "아르헨티나": "부에노스아이레스"
}

# 국가명 입력받기
country = input("국가명을 입력하세요: ").strip()

# 수도 출력하기
if country in capitals:
    print(f"{country}의 수도는 {capitals[country]}입니다.")
else:
    print("등록되지 않은 국가입니다.")
```
